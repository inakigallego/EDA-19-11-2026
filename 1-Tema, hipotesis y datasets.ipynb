# TEMA

Análisis empírico del Balance Global del Carbono (Global Carbon Budget), evaluando la evolución histórica de las concentraciones atmosféricas de CO2 y distinguiendo cuantitativamente el impacto de las emisiones antropogénicas frente a los flujos y sumideros naturales (océanos y ecosistemas terrestres). La atmósfera se aborda bajo un enfoque de contabilidad de masas ("caja de balance"), donde se acumula de forma neta el carbono que los sumideros planetarios no logran reabsorber.

# HIPOTESIS

La teoría científica y los modelos del ciclo del carbono establecen que el incremento continuado del CO2 atmosférico está impulsado de forma dominante por las emisiones antropogénicas (combustibles fósiles y cambios de uso del suelo), superando la capacidad dinámica de los sumideros naturales. Este estudio contrastará empíricamente dicha premisa mediante el cruce analítico de los registros históricos de la NOAA (National Oceanic and Atmospheric Administration) y las series globales de emisiones humanas recogidas por la OWID (Our World in Data).
Qué busco?
- Determinar qué porcentaje del dióxido de carbono que emitimos cada año se queda efectivamente en la atmósfera.
- Evaluar si la capacidad de absorción del planeta está aumentando al mismo ritmo que las emisiones o si muestra signos de saturación o pérdida de eficiencia.

# El factor de conversión físico: De masa a concentración (ppm)
La atmósfera de la Tierra tiene una masa total conocida. Gracias a ello, existe una equivalencia fija a nivel global para transformar gigatoneladas de carbono o dióxido de carbono en aumento de concentración atmosférica:
- 1 parte por millón (ppm) de CO2 en la atmósfera equivale aproximadamente a 2.13 gigatoneladas de carbono (GtC), o lo que es lo mismo, a 7.82 gigatoneladas de dióxido de carbono (GtCO2).
- Por lo tanto, para saber cuántas ppm teóricos subiría el CO2 si todo el gas emitido se quedara en el aire, se divide las toneladas de OWID entre 7.820.000.000 toneladas.

```python
df5_mundial = df4_mundial
columnas = [col for col in df5_mundial.columns if col not in ("country", "year")]
df5_mundial[columnas] = df5_mundial[columnas].div(7820)
df5_mundial
```
